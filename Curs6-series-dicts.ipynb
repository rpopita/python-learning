```python
# ============================================
# Exercițiul 3: Dicționar "tip DataFrame" - situatia notelor
# ============================================
# Cerinte:
#  a) gaseste toate notele unui student dupa nume
#  b) calculeaza media generala a clasei
#  c) afiseaza studentii sortati descrescator dupa media proprie

catalog = {
    "nume":      ["Ana", "Bogdan", "Carmen", "Dan", "Elena"],
    "note_mate": [9, 7, 10, 6, 8],
    "note_info": [10, 6, 9, 7, 9],
}


#  a) gaseste toate notele unui student dupa nume

nume = "Bogdan"
lista_nume = catalog["nume"]
indx_cautat = lista_nume.index(nume)
print("Indexul este: ", str(indx_cautat))
nota_mate = catalog["note_mate"][indx_cautat]
nota_info = catalog["note_info"][indx_cautat]

for i, value in enumerate(lista_nume):
    if value == nume:
        nota_mate = catalog["note_mate"][i]
        nota_info = catalog["note_info"][i]
        print(f"Pentru numele {nume}, avem notele: {nota_mate}, {nota_info}")
        break;
        

# for key, value in catalog.items():

#  b) calculeaza media generala a clasei
toate_notele = catalog["note_mate"] + catalog["note_info"]
print(toate_notele)
media = sum(toate_notele)/ len(toate_notele)
print(media)


#  c) afiseaza studentii sortati descrescator dupa media proprie
mediile_calculate = []
for i, value in enumerate(lista_nume):
    media_notelor = (catalog["note_mate"][i] + catalog["note_info"][i])/2
    mediile_calculate.append((value, media_notelor))

# print(mediile_calculate)
mediile_calculate.sort(key = lambda item: item[1], reverse = True)
# mediile_calculate.sort()
print(mediile_calculate)


lista= [3,7,9,1,2,4]
lista.sort()
print(lista)
```

```python
# ============================================
# Exercițiul 4: Curatarea tipurilor de date - Date calendaristice
# ============================================
# Avem date scrise in formate inconsistente (string). Cerinte:
#  a) converteste toate datele intr-un obiect datetime valid
#  b) sorteaza-le cronologic

from datetime import datetime

date_brute = ["2024-01-15", "15/03/2024", "2024.02.20", "01-04-2024"]

# folositi datetime.strptime(data_text, fmt), unde fmt = "%Y-%m-%d"

#  a) converteste toate datele intr-un obiect datetime valid

fmt_list = ["%Y-%m-%d", "%Y.%m.%d", "%d/%m/%Y", "%d-%m-%Y"]

def format_data(item):
    for fmt in fmt_list:
        try:
            return datetime.strptime(item, fmt)
        except ValueError:
            continue
    raise ValueError("Format neidentificat")

date_procesate = [ format_data(item) for item in date_brute]

print(date_procesate)

#  b) sorteaza-le cronologic

date_sortate = sorted(date_procesate)
print(date_sortate)
```
