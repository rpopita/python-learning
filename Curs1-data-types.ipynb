```python
## Liste 

fructe = ['măr', 'banană', 'portocală', 'kiwi', 'mango', 'pepene']

# slicing

print(fructe[0:]) # [0,2)
print(fructe[-1])
print(fructe[-3:])

print(fructe.index("portocală"))

# modificare lista - append/pop
fructe.append("lamaie")
print(fructe)


extragere_element = fructe.pop()
print(extragere_element)
print(fructe)

extragere_element2 = fructe.pop(0)
print(extragere_element2)
print(fructe)

# REMOVE

fructe.remove("banană")
print(fructe)

# SPLIT

prop = "Cursul,de,python"
print(prop.split(","))


lista_hibrida = [1,"unu", 2, "doi"]
print(lista_hibrida)

# JOIN

lista_reversed_to_string = ['Cursul', 'de', 'python']
prop_reversed = (" ").join(lista_reversed_to_string)
print(prop_reversed)


# sort

lista_numere = [5,6,1,2,3,10]
lista_numere.sort()
print(lista_numere)
lista_numere.sort(reverse=True)
print(lista_numere)

# insert
lista_numere = [5,6,1,2,3,10]
lista_numere.insert(2, 11)
print(lista_numere)


# functii similare cu sql

lista_numere = [5,6,1,2,3,10, 3, 6, 6, 7, 2, 1]
print(lista_numere.count(6))
print(min(lista_numere))
print(sum(lista_numere))
```
