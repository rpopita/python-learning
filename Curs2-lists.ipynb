```python
#shallow copy
list1 = [1,2,3]
list2 = list1
print(f"Is list 1 the same object as list 2? -> {list1 is list2}")

list2.append(4)

print(list1)
print(list2)
```

```python
# deep copy
list1 = [1,2,3]
list2 = list1.copy()
print(f"Is list 1 the same object as list 2? -> {list1 is list2}")

list2.append(4)

print(list1)
print(list2)
```

```python
# deep copy pe un singur nivel
list3 = [[1,2,3], [4,5]]
list4 = list3.copy()

list4.append(6)


print(list3)
print(list4)

#shallow copy pe nivelul 2

list4[0].append(3)
print("\n")
print(list3)
print(list4)

# deep copy complet
import copy

list5 = [[1,2,3], [4,5]]
list6 = copy.deepcopy(list5)

list6[0].append(3)

print("\n")
print(list5)
print(list6)
```

- Sa zicem ca vine black friday si avem datele lunare despre evolutia pretului unui singur produs. 
- Trebuie sa stochez aceste informatii intr-o lista.
- Am 13 luni, cu pret/luna si incepe din oct 2025 pana in oct 2026. Trebuie sa aflu pretul pentru nov 2026.
- presupunem ca pretul va creste in luna nov 2026 fata de oct 2026 cu acelasi procent cu care crescuse pretul initial din oct2025 pana in nov 2025

```python
prices = [100, 120, 115, 110, 125, 130, 135, 140, 145, 150, 155, 160, 165]
        # oct2025, nov2025, ....., oct2026

oct_2025 = prices[0]
nov_2025 = prices[1]

procent = (nov_2025 - oct_2025)/oct_2025 * 100
print(f"Procentul cu care a crescut pretul: {procent}%")

print(len(prices))
oct_2026 = prices[len(prices)-1]
nov_2026 = oct_2026 + oct_2026 * procent/100

print(f"Pretul pentru nov 2026: {nov_2026}RON")

# prices.append(nov_2026)
prices.extend([nov_2026])
print(prices)
```

## Mini Proiect personal - Calculator cheltuieli personale

### Obiectiv: 
Să scrieți un program în Python care simulează un gestionar de cheltuieli personale, rulat din consolă printr-un meniu interactiv.

### Cerințe funcționale

Programul trebuie să mențină o listă de cheltuieli, unde fiecare cheltuială este reprezentată printr-un dicționar cu următoarele chei:

- categorie (ex: "mancare", "transport", "distractie")
- suma (număr, în RON)
- descriere (text, opțional)


Programul trebuie să afișeze un meniu cu următoarele opțiuni, care se repetă până când utilizatorul alege să iasă:

1. Adaugă cheltuială — cere categorie, sumă și descriere de la utilizator și le adaugă în listă.
2. Afișează toate cheltuielile — listează toate cheltuielile înregistrate, cu index.
3. Filtrează după categorie — utilizatorul introduce o categorie, iar programul afișează doar cheltuielile din acea categorie.

5. Total și medie — calculează și afișează suma totală a cheltuielilor și media lor.

## LIST COMPREHENSION
https://claude.ai/public/artifacts/48718cfc-8027-40d4-af55-f3587a5a357a 

```python
#[expression for item in list if condition]
lista = [1,2,3,4,5,6,7,8]
squares = [ item**2 for item in lista if item % 2 == 0 ]
print(squares)


celsius = [0,10,20,30,40]
# F = C * 9/5 + 32
fahrenheit = [ c*9/5+32 for c in celsius]
print(fahrenheit)


negative_numbers = [-1, -2, -3, -4]
positives = [abs(item) for item in negative_numbers ]
print(positives)


data = [[1,2,3], [4,5,6], [7,8,9]]
lista_noua = [ [value*2 for value in rand] for rand in data]
print(lista_noua)
```

```python
# map
lista=[1,2,3,4,5,6]
list_c = [value *2 for value in lista]
print(list_c)
result_map = list(map(lambda value: value *2, lista))
print(result_map)

def triple_number(x):
    return x*3
result_map2 = list(map(triple_number, lista))
print(result_map2)
```

```python
# sort cu o conditie
words = ["banana", "kiwi", "portocala"]
words.sort(key = lambda w: len(w))
print(words)

# filter
lista=[1,2,3,4,5,6]
nr_pare = list(filter(lambda nr: nr%2==0, lista))
print(nr_pare)

# operatori ternari: conditie a < 5 ? a : 5
x = 10
rasp = "par" if x%2==0 else "impar"
print(rasp)

# op. ternari in list comprehension
numbers_categories= [f"impar: {x}" if x%2==1 else f"par: {x}" for x in lista]
print(numbers_categories)

# list c. with enumerate

fruits = ["banana", "kiwi", "portocala"]
indexed_fruits = [f"{index}: {fruit}" for index, fruit in enumerate(fruits)]
print(indexed_fruits)
```
