```python
import pandas as pd

serie1 = pd.Series([1,2,3])
print(serie1.astype(str).values)

# APPLY

print(serie1.apply(lambda x: x*2))

def modifica_val(x):
    return x*3

print(serie1.apply(modifica_val))
serie2 = serie1.apply(modifica_val)
print(serie2)

# list comprehension vs apply

rezultat_l_c = [ x*2 for x in serie1]
print(rezultat_l_c)
print(type(rezultat_l_c))
print(pd.Series(rezultat_l_c))


rezultat_apply = serie1.apply(lambda x: x*2)
print(type(rezultat_apply))
```

```python
### Exercițiul 6: Note cu valori lipsă

# O `Series` de note conține și valori lipsă (`None`). Cerințe:
# 1. Numărați câte note lipsesc.
# 2. Calculați media doar din notele existente.
# 3. Completați valorile lipsă cu media (`fillna`) și comparați lungimea cu varianta `dropna()`.

note = pd.Series([9, None, 7, None, 10, 8, None, 6, 7, None, None])

nr_lipsa = note.isna().sum()
print(nr_lipsa)

media= note.mean()


note_completate = note.fillna(media)
note_fara_lipsa = note.dropna()

print(note_completate)
print(note_fara_lipsa)

print(len(note))
print(len(note_completate))
print(len(note_fara_lipsa))
```

```python
### Exercițiul 8: Inventar de magazin

# Cerințe:
# 1. Din lista de dicționare `inventar`, construiți un `DataFrame`.
# 2. Folosind un `set`, aflați categoriile unice din magazin.
# 3. Calculați o `Series` cu valoarea totală de stoc (`preț × stoc`) per produs, sortată descrescător.
# 4. Construiți o `listă` cu primele 2 produse ca valoare de stoc.

inventar = [
    {"produs": "Laptop Gaming", "categorie": "Electronice", "pret": 4599.99, "stoc": 12},
    {"produs": "Mouse Wireless", "categorie": "Electronice", "pret": 89.90, "stoc": 45},
    {"produs": "Tricou Bumbac", "categorie": "Imbracaminte", "pret": 59.90, "stoc": 120},
    {"produs": "Blugi Slim", "categorie": "Imbracaminte", "pret": 149.00, "stoc": 30},
    {"produs": "Cafea Boabe", "categorie": "Alimentare", "pret": 39.90, "stoc": 200},
]

# 1. Din lista de dicționare `inventar`, construiți un `DataFrame`.
# 2. Folosind un `set`, aflați categoriile unice din magazin.

df_inventar = pd.DataFrame(inventar)
print(df_inventar)

categ_unice = set(df_inventar["categorie"])
print(categ_unice)

# 3. Calculați o `Series` cu valoarea totală de stoc (`preț × stoc`) per produs, sortată descrescător.
# 4. Construiți o `listă` cu primele 2 produse ca valoare de stoc.

valoare_stoc = (df_inventar["pret"] * df_inventar["stoc"]).sort_values(ascending=False)
print(valoare_stoc)
valoare_stoc.index = df_inventar["produs"]
print("\n")
print(valoare_stoc)


print(valoare_stoc.head(2))
print(list(valoare_stoc.index[:2]))
```
